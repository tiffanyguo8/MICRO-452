Finite state machine
1) Initialisation
- get the map created by global cam (WITHOUT INFLATION bc when obst detected we reinflate obst)
- get thymio's initial pos° (start)
- get the goal pos° (goal)

2) Planification
- do A*
- get the path
- smooth path if needed (already done by Role 3)
- give waypoints to controller

3) Moving
- Thymio follows waypoints
- pos° estim w sensors and odometry
- controller says when obst in the path or goal reached

4) Obstacle detected
- thymio detects new obst
- adapt trajectory to avoid imediate collision
- program get the info about the obst

5) Replanification
- actualize map
- get the actual estim pos°
- new A* to the goal
- send new path to controller

6) Mission done
- thymio reached its goal
- stop the motors
